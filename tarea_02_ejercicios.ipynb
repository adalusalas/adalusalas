```python
# ==============================================================================
# TAREA 02: VARIABLES, TIPOS DE DATOS Y CONDICIONALES
# Curso: Programación / Periodismo
# ==============================================================================

# ------------------------------------------------------------------------------
# PROBLEMA 1: Clasificador de temperatura
# ------------------------------------------------------------------------------
print("=== PROBLEMA 1: Clasificador de temperatura ===")

# Usamos un bucle for con range(-30, 51) para incluir desde el -30 hasta el 50
for temp in range(-30, 51):
    # Evaluamos si la temperatura es menor que 0
    if temp < 0:
        print(f"{temp} Muy frío")
    # Evaluamos si está entre 0 y 24
    elif 0 <= temp <= 24:
        print(f"{temp} Templado")
    # Si es 25 o mayor
    else:
        print(f"{temp} Caluroso")

print("\n" + "="*50 + "\n")


# ------------------------------------------------------------------------------
# PROBLEMA 2: Números especiales del robot
# ------------------------------------------------------------------------------
print("=== PROBLEMA 2: Números especiales del robot ===")

# Recorremos los números del 1 al 100 (range llega hasta 101 para incluir el 100)
for num in range(1, 101):
    # Verificamos primero si es divisible por 5 usando el operador módulo (%)
    if num % 5 == 0:
        # Evaluamos si ADEMÁS es divisible por 10
        if num % 10 == 0:
            print(f"{num} Bip Bip")
        else:
            print(f"{num} Bip")

print("\n" + "="*50 + "\n")


# ------------------------------------------------------------------------------
# PROBLEMA 3: Detector de puntaje gamer
# ------------------------------------------------------------------------------
print("=== PROBLEMA 3: Detector de puntaje gamer ===")

# Solicitamos los dos puntajes al usuario y los convertimos a enteros (int)
puntaje1 = int(input("Ingresa primer puntaje: "))
puntaje2 = int(input("Ingresa segundo puntaje: "))

# Calculamos la suma de los puntajes
suma_total = puntaje1 + puntaje2

# Clasificamos según las reglas establecidas
if suma_total < 50:
    print("\nResultado: Nivel principiante")
elif 50 <= suma_total < 80:
    print("\nResultado: Nivel intermedio")
else:
    print("\nResultado: Nivel pro")

print("\n" + "="*50 + "\n")


# ------------------------------------------------------------------------------
# PROBLEMA 4: Calculador de perfil musical
# ------------------------------------------------------------------------------
print("=== PROBLEMA 4: Calculador de perfil musical ===")

# Definimos los parámetros del usuario (puedes cambiar estos valores para probar)
edad = 20
le_gusta_musica = True
le_gusta_bailar = False

# Evaluamos las condiciones
if edad >= 18 and le_gusta_musica and le_gusta_bailar:
    print("Eres una estrella de la fiesta")
elif le_gusta_musica and not le_gusta_bailar:
    print("Te gusta la música, pero no tanto bailar")
else:
    print("Tienes tu propio estilo")
```
